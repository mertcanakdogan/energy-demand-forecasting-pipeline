# Synthetic energy-demand forecasting demo

This notebook is a lightweight exploration of the package APIs. The command-line pipeline remains the reproducible source of the committed example outputs. All observations below are generated locally from a fixed seed.

```python
from energy_forecasting.synthetic_data import SyntheticDataConfig, generate_synthetic_data
from energy_forecasting.validation import prepare_data, validate_data
from energy_forecasting.features import add_calendar_features

raw = generate_synthetic_data(
    SyntheticDataConfig(n_meters=4, n_groups=2, months=1, seed=42)
)
report = validate_data(raw)
report.to_frame()
```

```python
prepared = prepare_data(raw)
featured = add_calendar_features(prepared)
featured[["meter_id", "timestamp", "consumption_kwh", "temperature_c", "hour", "day_of_week"]].head()
```

```python
import matplotlib.pyplot as plt

example = prepared.query("meter_id == 'M0001'").tail(24 * 7)
example.plot(x="timestamp", y="consumption_kwh", figsize=(12, 4), title="One synthetic meter — final week")
plt.ylabel("Consumption (kWh)");
```

## Run the complete workflow

From the repository root, run `python -m energy_forecasting.pipeline`. The configured demo evaluates every enabled model for every meter, writes tabular outputs, and regenerates the charts displayed in the README.
